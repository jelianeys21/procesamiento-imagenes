```python

import numpy as np #Llamas numpy "as"
import matplotlib.pyplot as plt
```

```python
#Bandera de guatemala
Bandera = np.zeros((3, 3, 3))
Rojo =np.array([0,    255, 255])
Verde =np.array([255, 255,   255])
Azul = np.array([0,    255, 255])
Bandera[:][:][0] = Rojo
Bandera[:][:][1] = Verde
Bandera[:][:][2] = Azul
# transpose(posiciones del array 3d)
voltear = Bandera.transpose(1, 0, 2)
plt.imshow(voltear)
```

```python
# BANDERA DE EL PARAGUAY
Bandera = np.zeros((3, 3, 3))
Rojo = np.array([255, 0, 0])
Verde = np.array([255, 255, 255])
Azul =np.array([0, 0, 255])
Bandera[:][:][0] = Rojo
Bandera[:][:][1] = Verde
Bandera[:][:][2] = Azul
print(Bandera)
plt.imshow(Bandera)
```

```python
#EXTRAER IMAGEN RGB
import cv2
from google.colab.patches import cv2_imshow
!curl -o imagen https://definicionde.es/wp-content/uploads/2019/04/definicion-de-persona-min.jpg
img = cv2.imread("imagen")
imagen_gris = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2_imshow(imagen_gris)
umbral = 128
(_, imagen_binaria) = cv2.threshold(imagen_gris, umbral, 255, cv2.THRESH_BINARY)
cv2_imshow(imagen_binaria)
```

```python
#EXTRAER IMAGEN GRIS
!curl -o imagen2 https://images.wunderstock.com/Elephant-in-Africa_NLBWDLuRnYTO_400.jpeg
img2 = cv2.imread("imagen2")
umbral = 128
(_, imagen_binaria2) = cv2.threshold(img2, umbral, 255, cv2.THRESH_BINARY)
cv2_imshow(imagen_binaria2)
```

**PREGUNTA 3**

```python
#RECORTAR IMAGEN
(f, c) = imagen_gris.shape
d = len(imagen_gris.shape)
if d == 2:
  inicio_x = 200
  inicio_y = 50
  alto = 400
  ancho = 400
  imagen_cortada = imagen_gris[inicio_y:inicio_y+alto, inicio_x:inicio_x+ancho]
  cv2_imshow(imagen_cortada)
```

```python
#COLOCAR MARCO A LA IMAGEN RECORTADA
(f, c) = imagen_cortada.shape
d = len(imagen_cortada.shape)
if d == 2:
  alto = 50
  ancho = 50
  nueva_imagen = np.zeros((f+alto*2, c+ancho*2))
  nueva_imagen[alto:-alto, ancho:-ancho] = imagen_cortada
  cv2_imshow(nueva_imagen)
```
