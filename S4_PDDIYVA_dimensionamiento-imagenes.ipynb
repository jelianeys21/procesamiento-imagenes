```python
import cv2
from google.colab.patches import cv2_imshow
import numpy as np
import matplotlib.pyplot as plt
```

```python
# lectura de la imagen
imagen = cv2.imread("foto.jpg")
# dimensionamiento de la imagen
imagen_dimensionada = cv2.resize(imagen, (500, 300))
cv2_imshow(imagen_dimensionada)
```

```python
# Pasar la imagen de RGB a gris
imagen_gris = cv2.cvtColor(imagen_dimensionada, cv2.COLOR_BGR2GRAY)
cv2_imshow(imagen_gris)
```

**HISTOGRAMA**

```python
# cv2.calcHist(imagen, canal, mascara, tamano_histograma, rangos)
hist = cv2.calcHist([imagen_gris], [0], None, [256], [0, 255])
plt.plot(hist)
plt.xlabel("intensidad de iluminación")
plt.ylabel("cantidad de píxeles")
```

```python
# import matplotlib.pyplot as plt
# import numpy as np
# def imagen_histograma(imagen_gris):
(f, c) = imagen_gris.shape
histograma = np.zeros((1, 256))
# histograma[0]
for i in range(0, f):
  for j in range(0, c):
    valor = imagen_gris[i][j]
    histograma[0][valor] = histograma[0][valor] + 1
plt.plot(histograma[0])
```

**AGREGAR MARCOS**

```python
# import numpy as np
# def agregar_marcos(imagen_gris, alto, ancho):
(f, c) = imagen_gris.shape #500x500
d = len(imagen_gris.shape)
# si es una imagen gris o blanco y negro
if d == 2:
  alto = 50
  ancho = 100
  nueva_imagen = np.zeros((f+alto*2, c+ancho*2)) #700x700
  nueva_imagen[alto:-alto, ancho:-ancho] = imagen_gris
  # return nueva_imagen.astype(np.uint8)
  cv2_imshow(nueva_imagen)
```

**CORTAR IMAGENES**

```python
# import numpy as np
# def cortar_imagen(imagen_gris, inicio_x, inicio_y, alto, ancho):
(f, c) = imagen_gris.shape #500x500
d = len(imagen_gris.shape)
# si es una imagen gris o blanco y negro
if d == 2:
  inicio_x = 200
  inicio_y = 10
  alto = 200
  ancho = 200
  imagen_cortada = imagen_gris[inicio_y:inicio_y+alto, inicio_x:inicio_x+ancho]
  # return imagen_cortada.astype(np.uint8)
  cv2_imshow(imagen_cortada)
```

**IMAGEN REFLEJADA**

```python
# import numpy as np
# def reflejar_imagen(imagen_gris):
d = len(imagen_gris.shape)
if d == 2:
  (f, c) = imagen_gris.shape #500x500
  # si es una imagen gris o blanco y negro
  imagen_reflejada = np.zeros((f, c)) #500x500
  for i in range(0, f):
    for j in range(0, c):
      imagen_reflejada[i][j] = imagen_gris[i, c-(j+1)]
  # return imagen_reflejada.astype(np.uint8)
  cv2_imshow(imagen_reflejada)
else:
  print("No es una imagen de 2 dimensiones")
```
