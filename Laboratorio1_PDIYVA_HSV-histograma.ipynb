**PREGUNTA 1**

```python
paisaje = cv2.imread('/content/curso_pdi/imagenes/laboratorio1/paisaje.jpg')
AzulBajo1 = np.array([95, 0, 0], np.uint8)
AzulAlto1 = np.array([135, 255, 255], np.uint8)
imagen_hsv = cv2.cvtColor(paisaje, cv2.COLOR_BGR2HSV)
mascaraAzul = cv2.inRange(imagen_hsv, AzulBajo1, AzulAlto1)
deteccionAzul = cv2.bitwise_and(paisaje, paisaje, mask=mascaraAzul)
cv2_imshow(deteccionAzul)
```

**PREGUNTA 4**

```python
persona = cv2.imread('/content/curso_pdi/imagenes/laboratorio1/persona_lentes.jpg')
persona_gris = cv2.cvtColor(persona, cv2.COLOR_BGR2GRAY)
(f, c) = persona_gris.shape
factor_correccion = 70
persona_brillo = np.zeros((f, c))
for i in range(0, f):
  for j in range(0, c):
    persona_brillo[i][j] = persona_gris[i][j] + factor_correccion
    if persona_brillo[i][j] > 255:
      persona_brillo[i][j] = 255
    elif persona_brillo[i][j] < 0:
      persona_brillo[i][j] = 0
persona_brillo = persona_brillo.astype(np.uint8)
persona_contraste = np.zeros((f, c))
maximo = persona_gris.max()
minimo = persona_gris.min()
for i in range(0, f):
  for j in range(0, c):
    pixel_actual = persona_gris[i][j]
    persona_contraste[i][j] = round(((pixel_actual - minimo) / (maximo - minimo)) * 255)
    persona_contraste[i][j] = persona_contraste[i][j] + 50
    if persona_contraste[i][j] > 255:
      persona_contraste[i][j] = 255
    elif persona_contraste[i][j] < 0:
      persona_contraste[i][j] = 0
persona_contraste = persona_contraste.astype(np.uint8)
persona_final = cv2.add(persona_brillo, persona_contraste)
cv2_imshow(persona_gris)
cv2_imshow(persona_final)
```
